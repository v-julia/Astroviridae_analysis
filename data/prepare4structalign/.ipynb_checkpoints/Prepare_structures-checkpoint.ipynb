```python
import os
from Bio.PDB import MMCIFParser, PDBIO

import shutil
import glob, json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def cif2pdb(mcif_path, pdb_path):
    parser = MMCIFParser(QUIET=True)
    structure = parser.get_structure("protein", mcif_path)
    io = PDBIO()
    io.set_structure(structure)
    io.save(pdb_path)
```

# Visualise quality of prediction for all capsid proteins

```python
BACKBONE = {"N", "CA", "C", "O"}
HIGH, CONF, LOW = 90, 70, 50

parser = MMCIFParser(QUIET=True)


def find_mcif_json_pairs(root):
    """Return (entry_id, json_path, cif_path) for each AF3 result folder."""
    for batch_dir in sorted(os.listdir(root)):
        bpath = os.path.join(root, batch_dir)
        if not os.path.isdir(bpath):
            continue
        for entry_dir in sorted(os.listdir(bpath)):
            epath = os.path.join(bpath, entry_dir)
            if not os.path.isdir(epath):
                continue

            parts = entry_dir.split("_")
            entry_id = "_".join(parts[1:]) if len(parts) > 2 else (parts[1] if len(parts) == 2 else entry_dir)

            # Find full-data JSON (server or local naming)
            jsons = glob.glob(os.path.join(epath, "*_full_data_*.json"))
            if not jsons:
                print(f"[skip] no full-data JSON in {epath}")
                continue

            # Find model CIF
            cif = os.path.join(epath, f"fold_{entry_dir}_model_0.cif")
            if not os.path.exists(cif):
                print(f"[skip] no CIF in {epath}")

            yield entry_id, jsons[0], cif


def residue_plddt_from_json_and_cif(json_path, cif_path):
    """
    Map atom_plddts from JSON to residues using atom order from CIF.
    Returns DataFrame: chain, resseq, icode, resname, mean/min/max pLDDT, n_atoms.
    """
    # 1. Read per-atom pLDDT and chain IDs from JSON
    with open(json_path) as fh:
        d = json.load(fh)
    plddt = np.asarray(d["atom_plddts"], dtype=float)
    json_chains = d.get("atom_chain_ids", [])

    # 2. Parse CIF to get atom metadata in the same order
    structure = parser.get_structure("model", cif_path)
    atoms_meta = []
    for model in structure:
        for chain in model:
            for residue in chain:
                for atom in residue:
                    atoms_meta.append({
                        "chain": chain.id,
                        "resseq": residue.id[1],
                        "icode": residue.id[2].strip(),
                        "resname": residue.resname,
                        "atom": atom.get_name(),
                        "plddt": None,   # fill below
                    })
        break  # only first model

    # 3. Check length match
    if len(atoms_meta) != len(plddt):
        raise ValueError(
            f"Atom count mismatch: CIF has {len(atoms_meta)}, JSON has {len(plddt)} in {json_path}"
        )

    # 4. Assign pLDDT by atom index
    for meta, val in zip(atoms_meta, plddt):
        meta["plddt"] = val

    df = pd.DataFrame(atoms_meta)

    # 5. Restrict to backbone atoms for per-residue pLDDT
    bb = df[df["atom"].isin(BACKBONE)]
    if bb.empty:
        bb = df  # fallback: use all atoms

    agg = (bb.groupby(["chain", "resseq", "icode", "resname"])
             .agg(mean_plddt=("plddt", "mean"),
                  min_plddt=("plddt", "min"),
                  max_plddt=("plddt", "max"),
                  n_atoms=("plddt", "count"))
             .reset_index())
    return agg
```

```python
OUT_DIR = "D:/Virology/RNA_viruses/Astrovirus/Astroviridae_analysis/data/capsid_prediction"
records = []
for entry_id, json_path, cif_path in find_mcif_json_pairs(mcif_folder):
    try:
        df = residue_plddt_from_json_and_cif(json_path, cif_path)
    except Exception as e:
        print(f"[error] {entry_id}: {e}")
        continue
    df.insert(0, "entry", entry_id)
    records.append(df)
    print(f"parsed {entry_id}: {len(df)} residues, mean pLDDT={df.mean_plddt.mean():.1f}")

if not records:
    raise SystemExit("No results parsed.")

per_residue = pd.concat(records, ignore_index=True)
per_residue.to_csv(os.path.join(OUT_DIR, "af3_per_residue_plddt.csv"), index=False)
```

```python
# ---------------------------------------------------------------------------
# Per-entry summary
# ---------------------------------------------------------------------------
per_entry = (per_residue.groupby("entry")
             .agg(n_residues=("resseq", "count"),
                  mean_plddt=("mean_plddt", "mean"),
                  median_plddt=("mean_plddt", "median"),
                  min_plddt=("min_plddt", "min"),
                  frac_high=("mean_plddt", lambda s: (s >= HIGH).mean()),
                  frac_conf=("mean_plddt", lambda s: (s >= CONF).mean()),
                  frac_low=("mean_plddt", lambda s: (s < LOW).mean()))
             .sort_values("mean_plddt", ascending=False))
per_entry.to_csv(os.path.join(OUT_DIR, "af3_capsid_per_entry_plddt.csv"))

# ---------------------------------------------------------------------------
# Per-position summary across entries
# ---------------------------------------------------------------------------
per_position = (per_residue.groupby(["chain", "resseq", "icode"])
                .agg(n_entries=("entry", "nunique"),
                     mean_across=("mean_plddt", "mean"),
                     median_across=("mean_plddt", "median"),
                     std_across=("mean_plddt", "std"),
                     min_across=("mean_plddt", "min"),
                     max_across=("mean_plddt", "max"))
                .reset_index())
per_position.to_csv(os.path.join(OUT_DIR, "af3_capsid_per_position_plddt.csv"), index=False)

conf_frac = (per_residue.assign(conf=per_residue.mean_plddt >= CONF)
             .groupby(["chain", "resseq", "icode"])["conf"]
             .agg(n_confident="sum", frac_confident="mean")
             .reset_index())
conf_frac.to_csv(os.path.join(OUT_DIR, "af3_capsid_per_position_confident_fraction.csv"), index=False)
```

```python
# ---------------------------------------------------------------------------
# Plot: pLDDT per position for all structures
# ---------------------------------------------------------------------------
for chain in per_position["chain"].unique():
    pos = per_position[per_position.chain == chain].sort_values(["resseq", "icode"])
    fig, ax = plt.subplots(figsize=(14, 4))

    for entry, sub in per_residue[per_residue.chain == chain].groupby("entry"):
        sub = sub.sort_values(["resseq", "icode"])
        x = sub.resseq + sub.icode.apply(lambda c: (ord(c) - 65) / 10 if c else 0)
        ax.plot(x, sub.mean_plddt, color="grey", lw=0.4, alpha=0.35)

    x_mean = pos.resseq + pos.icode.apply(lambda c: (ord(c) - 65) / 10 if c else 0)
    ax.plot(x_mean, pos.mean_across, color="black", lw=1.6,
            label=f"mean (n={pos.n_entries.iloc[0]})")

    ax.axhline(HIGH, color="green", ls="--", lw=0.8)
    ax.axhline(CONF, color="orange", ls="--", lw=0.8)
    ax.axhline(LOW, color="red", ls="--", lw=0.8)
    ax.set_ylim(0, 100)
    ax.set_xlabel("Residue position")
    ax.set_ylabel("pLDDT (backbone mean)")
    ax.set_title(f"Chain {chain}: per-position pLDDT across {pos.n_entries.iloc[0]} structures")
    ax.legend(loc="lower right", fontsize=8)
    plt.tight_layout()
    out = os.path.join(OUT_DIR, f"af3_capsid_plddt_per_position_chain{chain}.png")
    plt.savefig(out, dpi=200)
    plt.close()
    print(f"wrote {out}")
```

```python
# ----- (5) heatmap: structure x position -----
mat = per_residue.pivot_table(index="entry", columns="resseq", values="mean_plddt")
mat = mat.loc[mat.mean(axis=1).sort_values(ascending=False).index]
fig, ax = plt.subplots(figsize=(14, max(4, mat.shape[0] * 0.08)))
im = ax.imshow(mat.values, aspect="auto", cmap="RdYlGn",
               vmin=0, vmax=100, interpolation="nearest")
ax.set_xlabel("Residue position"); ax.set_ylabel("Structure (by mean pLDDT)")
ax.set_title("AF2 ORF1b: pLDDT per position x structure")
if mat.shape[0] > 40:
    step = max(1, mat.shape[0] // 20)
    ax.set_yticks(range(0, mat.shape[0], step))
    ax.set_yticklabels(mat.index[::step], fontsize=6)
plt.colorbar(im, ax=ax, fraction=0.02, pad=0.01, label="pLDDT")
plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR, "af3_capsid_plddt_heatmap.png"), dpi=180)
plt.close()
```

# Plot per-residue pLDDT

```python
BATCHES_DIR = "D:/Virology/RNA_viruses/Astrovirus/Structure_prediction/ORF1b/af2_batches"
OUT_DIR     = "D:/Virology/RNA_viruses/Astrovirus/Astroviridae_analysis/data/predicted_structuresORF1b_83nt"

SCORES_GLOB = "*_scores_rank_001_*.json"
HIGH, CONF, LOW = 90, 70, 50

# ---------------------------------------------------------------------------
# Same entry-ID logic as your copy script
# ---------------------------------------------------------------------------
def entry_id_from_filename(fname):
    parts = fname.split("_")
    if parts[0] in ("NC", "YP", "NP"):
        return parts[0] + "_" + parts[1]      # e.g. "NP_056779.1"
    return parts[0]                            # e.g. "YP529897.1" style IDs

# ---------------------------------------------------------------------------
# Parse one scores JSON -> per-residue pLDDT
# ---------------------------------------------------------------------------
def residue_plddt_from_scores(json_path):
    with open(json_path) as fh:
        d = json.load(fh)
    plddt = d.get("plddt")
    if plddt is None:
        raise KeyError(f"'plddt' not found in {json_path}")
    plddt = np.asarray(plddt, dtype=float)
    return pd.DataFrame({
        "resseq":     np.arange(1, len(plddt) + 1),
        "mean_plddt": plddt,
    })

# ---------------------------------------------------------------------------
# Discover one best-model JSON per entry across all batch subfolders
# ---------------------------------------------------------------------------
json_files = sorted(glob.glob(os.path.join(BATCHES_DIR, "*", SCORES_GLOB)))
print(f"{len(json_files)} scores JSON(s) across batch subfolders")

records = []
seen = {}                              # entry_id -> json path (catch duplicates)
for jf in json_files:
    fname  = os.path.basename(jf)
    entry  = entry_id_from_filename(fname)
    if entry in seen:
        print(f"  [duplicate] {entry}: keeping {seen[entry]}, ignoring {jf}")
        continue
    seen[entry] = jf

    try:
        df = residue_plddt_from_scores(jf)
    except Exception as e:
        print(f"  [error] {entry}: {e}")
        continue
    df.insert(0, "entry", entry)
    records.append(df)

per_residue = pd.concat(records, ignore_index=True)
per_residue.to_csv(os.path.join(OUT_DIR, "af2_orf1b_per_residue_plddt.csv"), index=False)
print(f"\nParsed {per_residue.entry.nunique()} entries, "
      f"{len(per_residue)} residue rows")
```

```python
per_entry = (per_residue.groupby("entry")["mean_plddt"]
             .agg(n_residues="count",
                  mean="mean", median="median", std="std",
                  min="min", max="max",
                  frac_high=lambda s: (s >= HIGH).mean(),
                  frac_conf=lambda s: (s >= CONF).mean(),
                  frac_low =lambda s: (s <  LOW ).mean())
             .sort_values("mean", ascending=False))
per_entry.to_csv(os.path.join(OUT_DIR, "af2_per_entry_plddt.csv"))
print(f"per_entry_plddt.csv: {len(per_entry)} entries")

# ----- (2) per-position summary -----
group_cols = [c for c in ("chain", "resseq", "icode") if c in per_residue.columns]
per_position = (per_residue.groupby(group_cols)["mean_plddt"]
                .agg(n_entries="count",
                     mean_across="mean",
                     median_across="median",
                     std_across="std",
                     min_across="min",
                     max_across="max",
                     q25=lambda s: s.quantile(0.25),
                     q75=lambda s: s.quantile(0.75))
                .reset_index())
per_position.to_csv(os.path.join(OUT_DIR, "af2_per_position_plddt.csv"), index=False)

# ----- (3) per-position confidence fraction -----
conf_frac = (per_residue.assign(conf=per_residue.mean_plddt >= CONF)
             .groupby(group_cols)["conf"]
             .agg(n_confident="sum", frac_confident="mean")
             .reset_index())
conf_frac.to_csv(os.path.join(OUT_DIR, "af2_per_position_confident_fraction.csv"),
                 index=False)

# ----- (4) plot: thin line per structure + bold mean -----
fig, ax = plt.subplots(figsize=(14, 4))
for entry, sub in per_residue.groupby("entry"):
    sub = sub.sort_values("resseq")
    ax.plot(sub.resseq, sub.mean_plddt, color="grey", lw=0.4, alpha=0.30)

pos = per_position.sort_values("resseq")
ax.plot(pos.resseq, pos.mean_across, color="black", lw=1.6,
        label=f"mean (n={pos.n_entries.iloc[0]})")
ax.axhline(HIGH, color="green",  ls="--", lw=0.8)
ax.axhline(CONF, color="orange", ls="--", lw=0.8)
ax.axhline(LOW,  color="red",    ls="--", lw=0.8)
ax.set_ylim(0, 100)
ax.set_xlabel("Residue position"); ax.set_ylabel("pLDDT")
ax.set_title(f"AF2 ORF1b: pLDDT per position across {per_entry.shape[0]} structures")
ax.legend(loc="lower right", fontsize=8)
plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR, "af2_plddt_per_position.png"), dpi=200)
plt.close()

# ----- (5) heatmap: structure x position -----
mat = per_residue.pivot_table(index="entry", columns="resseq", values="mean_plddt")
mat = mat.loc[mat.mean(axis=1).sort_values(ascending=False).index]
mat.to_csv(os.path.join(OUT_DIR, "af2_per_residue_plddt_heatmap.csv"))
fig, ax = plt.subplots(figsize=(14, max(4, mat.shape[0] * 0.08)))
im = ax.imshow(mat.values, aspect="auto", cmap="RdYlGn",
               vmin=0, vmax=100, interpolation="nearest")
ax.set_xlabel("Residue position"); ax.set_ylabel("Structure (by mean pLDDT)")
ax.set_title("AF2 ORF1b: pLDDT per position x structure")
if mat.shape[0] > 40:
    step = max(1, mat.shape[0] // 20)
    ax.set_yticks(range(0, mat.shape[0], step))
    ax.set_yticklabels(mat.index[::step], fontsize=6)
plt.colorbar(im, ax=ax, fraction=0.02, pad=0.01, label="pLDDT")
plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR, "af2_plddt_heatmap.png"), dpi=180)
plt.close()

# ----- (6) violin: distribution per position -----
positions = sorted(per_residue["resseq"].unique())
fig, ax = plt.subplots(figsize=(max(12, len(positions) * 0.22), 4))
sns.violinplot(data=per_residue, x="resseq", y="mean_plddt",
               order=positions, cut=0, inner="quartile",
               linewidth=0.3, ax=ax, color="steelblue")
ax.axhline(HIGH, color="green",  ls="--", lw=0.8)
ax.axhline(CONF, color="orange", ls="--", lw=0.8)
ax.axhline(LOW,  color="red",    ls="--", lw=0.8)
ax.set_ylim(0, 100)
ax.set_xlabel("Residue position"); ax.set_ylabel("pLDDT")
ax.set_title("AF2 ORF1b: distribution of pLDDT per position")
if len(positions) > 60:
    ax.set_xticks(positions[:: max(1, len(positions) // 30)])
plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR, "af2_plddt_violin.png"), dpi=180)
plt.close()

print("Done.")
```

```python
capsid_ids = [x.split('.')[0].upper() for x in os.listdir(capsid_path)]

pol_files = os.listdir(pol_path)
astro_bastro = 0
astro_bastro_hpc = 0
astro_bastro_ids = []
for pol_file in pol_files:
    pol_id = pol_file.split(".")[0]

    if pol_id in capsid_ids:
        astro_bastro_ids.append(pol_id)
        astro_bastro +=1
        astro_bastro_hpc +=1
        src = os.path.join(pol_path, pol_file)
        dst1 = os.path.join(pol_astrobastro_path, pol_file)
        dst2 = os.path.join(pol_redbastro_path, pol_file)
        shutil.copy2(src, dst1)
        shutil.copy2(src, dst2)
    else:
        print(pol_id)
    if pol_id.startswith("YP") or pol_id.startswith("NP"):
        astro_bastro_hpc+=1
        src = os.path.join(pol_path, pol_file)
        dst = os.path.join(pol_redbastro_path, pol_file)
        shutil.copy2(src, dst)
        
print(f"{astro_bastro} entries were copied to {pol_astrobastro_path}")
print(f"{astro_bastro_hpc} entries were copied to {pol_redbastro_path}")
```
