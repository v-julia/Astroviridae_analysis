```python
from Bio import SeqIO
import os
```

```python
def create_rec_name_dict(alignments):
    rec_id_dict = {}
    for alignment in alignments:
        recs = SeqIO.parse(alignment, "fasta")
        for rec in recs:
            rec_id = rec.name.split("/")[0].split("|")[0]
            rec_id_dict[rec_id] = rec.name
    return(rec_id_dict)

def rename_usalignment(usalig_folder, out_dir, rec_id_dict):
    
    us_alignments = list(filter(lambda x: ".fasta" in x, os.listdir(usalig_folder)))
    for alignment in us_alignments:
        recs = list(SeqIO.parse(os.path.join(usalig_folder,alignment), "fasta"))
        for rec in recs:
            rec_name_list = rec.name.split(".")
            if rec_name_list[1] == "1":
                rec_id = ".".join(rec_name_list[0:2]).upper()
            else:
                rec_id = rec_name_list[0].upper()
            if rec_id in rec_id_dict.keys():
                rec.id = rec_id_dict[rec_id]
                rec.name = rec_id_dict[rec_id]
                rec.description = ""
        SeqIO.write(recs, os.path.join(out_dir,alignment), "fasta")
```

```python
usalig_foldtree = "D:/Virology/RNA_viruses/Astrovirus/Astroviridae_analysis/data/usalign"

out_dir_root = "D:/Virology/RNA_viruses/Astrovirus/Astroviridae_analysis/data/usalign/renamed"

alignments = ["D:/Virology/RNA_viruses/Astrovirus/Astroviridae_analysis/data/orf1b_83nt_for_prediction/Astroviridae_complete_1B_nt83_aa70_HPC.faa",
             "D:/Virology/RNA_viruses/Astrovirus/Astroviridae_analysis/data/bastrovirus_dataset/cds_assigned/bastro_rdrp_reduced.fasta"]


dict_rec_id = create_rec_name_dict(alignments)
rename_usalignment(usalig_foldtree, out_dir_root,dict_rec_id)
```
