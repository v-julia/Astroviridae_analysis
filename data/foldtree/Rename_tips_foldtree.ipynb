```python
from Bio import SeqIO
import os
```

```python
def create_rec_name_dict(alignments):
    rec_id_dict = {}
    for alignment in alignments:
        recs = SeqIO.parse(alignment, "fasta")
        for rec in recs:
            rec_id = rec.name.split("/")[0].split("|")[0]
            rec_id_dict[rec_id] = rec.name
    return(rec_id_dict)
```

```python
def rename_foldtree_trees(folder_foldtree, out_dir, alignments):
    rec_id_dict = create_rec_name_dict(alignments)
    trees = list(filter(lambda x: ".nwk" in x, os.listdir(folder_foldtree)))
    for tree_file in trees:
        with open(os.path.join(folder_foldtree, tree_file)) as tree:
            tree_line = "".join(tree.readlines()).upper()
            tree_line_new = tree_line
            for recid, recname in rec_id_dict.items():
                tree_line_new = tree_line_new.replace(recid, recname)
        tree.close()
        with open(os.path.join(out_dir, tree_file), "w") as new_tree_file:
            new_tree_file.write(tree_line_new)
        new_tree_file.close()
        print(f"{os.path.join(out_dir, tree_file)} was written")
```

```python
foldtree_folder_root = "D:/Virology/RNA_viruses/Astrovirus/Structure_prediction/ORF1b"
foldtree_suffixes = ["af2_pdb_astrobastro", "af2_pdb_redbastro","af2_pdb"]

out_dir_root = "D:/Virology/RNA_viruses/Astrovirus/Astroviridae_analysis/data/foldtree"

alignments = ["D:/Virology/RNA_viruses/Astrovirus/Astroviridae_analysis/data/orf1b_83nt_for_prediction/Astroviridae_complete_1B_nt83_aa70_HPC.faa",
             "D:/Virology/RNA_viruses/Astrovirus/Astroviridae_analysis/data/bastrovirus_dataset/cds_assigned/bastro_rdrp_reduced.fasta"]




for foldtree_suffix in foldtree_suffixes:
    foldtree_folder = os.path.join(foldtree_folder_root, foldtree_suffix)
    out_dir = os.path.join(out_dir_root, foldtree_suffix)
    try:
        os.mkdir(out_dir)  
    except:
        print(f"was not able to make dir {out_dir}")
    rename_foldtree_trees(foldtree_folder, out_dir, alignments)
```
