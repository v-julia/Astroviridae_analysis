```python
import json
import os
import pandas as pd
from Bio import SeqIO
```

```python
def create_json(path_fasta, path_out, max_length=4999, max_jobs = 30):
    records = SeqIO.parse(path_fasta, "fasta")
    batch = 1
    num_jobs = 0
    list_dicts=[]
    for rec in records:
        if num_jobs >= 30:
            print(len(list_dicts))
            with open(f"{path_out}/alphafold3_separate_jobs_{batch}.json", "w") as f:
                json.dump(list_dicts, f, indent=2)
            list_dicts = []
            num_jobs = 0
            batch+=1
        #if cum_length + len(rec.seq) < max_length:
        chain = {
                "proteinChain": {
                "sequence": str(rec.seq),
                "count": 1
              } }
        dict_job = {
                "name":  "rdrp " + rec.id.split("/")[0].split("|")[0].split(".")[0],
                "modelSeeds": [],
                "sequences": [chain],
                "dialect": "alphafoldserver",
                "version": 1}
        list_dicts.append(dict_job)
        num_jobs +=1
    
    if len(list_dicts) > 1:
        with open(f"{path_out}/alphafold3_separate_jobs_{batch}.json", "w") as f:
            json.dump(list_dicts, f, indent=2)
    
    print(f"Built {batch} batches")
```

```python
root = "D:/Virology/RNA_viruses/Astrovirus/Astroviridae_analysis/data/"
bastro_rdrp = f"{root}bastrovirus_dataset/cds_assigned/bastro_rdrp_noamb_reduced.fasta"
astro_hpc_rdrp = f"{root}orf1b_83nt_for_prediction/Astroviridae_complete_1B_nt83_aa70_HPC.faa"

alphafold_batches = f"{root}orf1b_83nt_for_prediction/alphafold3_batches"
```

```python
create_json(bastro_rdrp, alphafold_batches + "/bastro", max_length=4999, max_jobs = 30)
```

```python
create_json(astro_hpc_rdrp, alphafold_batches + "/astro_hpc", max_length=4999, max_jobs = 30)
```
